```python
## Text Splitting Techniques

### How to recursively split text by characters: library langchain-text-splitters

from langchain_text_splitters import RecursiveCharacterTextSplitter

speech=""
with open("speech.txt") as f:
    speech=f.read()


text_splitter = RecursiveCharacterTextSplitter(chunk_size=5,chunk_overlap=3)

final_documents = text_splitter.create_documents([speech])

type(final_documents[0])
```
